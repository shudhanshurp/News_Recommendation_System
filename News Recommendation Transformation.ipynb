```python
%pip install wordninja nltk --quiet
from pyspark.sql import SparkSession
import wordninja
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import numpy as np
from pyspark.sql.functions import lower, concat, regexp_replace, split, col, udf, size, concat_ws
from pyspark.sql.types import ArrayType, StringType, FloatType
import nltk
import ast
nltk.download('stopwords')
```

```python
news_df = spark.read.csv(
    "/mnt/newsrecsysdata/raw-data/news.tsv",
    sep='\t',
    header=False,
    inferSchema=True
).toDF('News ID', 'Category', 'SubCategory', 'Title', 'Abstract', 'URL', 'Title Entities', 'Abstract Entities')
display(news_df)
```

```python
news_df = news_df.dropna()
news_df = news_df.dropDuplicates()
```

```python
split_words = udf(lambda x: ' '.join(wordninja.split(x)), StringType())
news_df = news_df.withColumn('Category', split_words(news_df['Category']))
news_df = news_df.withColumn('SubCategory', split_words(news_df['SubCategory']))
news_df = news_df.withColumn('Category', lower(news_df['Category']))
news_df = news_df.withColumn('SubCategory', lower(news_df['SubCategory']))
news_df = news_df.withColumn('Title', lower(news_df['Title']))
news_df = news_df.withColumn('Abstract', lower(news_df['Abstract']))
display(news_df)
```

```python
def remove_stopwords(text):
    stopwords = set(nltk.corpus.stopwords.words('english'))
    words = text.split()
    filtered_words = [word for word in words if word.lower() not in stopwords]
    return ' '.join(filtered_words)

remove_stopwords_udf = udf(remove_stopwords, StringType())

news_df = news_df.withColumn('Content', concat(col('Category'), col('SubCategory'), col('Title'), col('Abstract')))
news_df = news_df.withColumn('Content', regexp_replace('Content', r'[^\w\s]', ''))
news_df = news_df.withColumn('Content', remove_stopwords_udf('Content'))
display(news_df)
```

```python
news_df = news_df.dropna()
display(news_df)
```
